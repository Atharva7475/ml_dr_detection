# 8. Reproducible Stratified Dataset Splitting (70% Train, 15% Val, 15% Test)

Splits the 3,662 APTOS samples into stratified Train (70%), Validation (15%), and Test (15%) subsets using `RANDOM_STATE = 42` and saves CSV files to `src/data/splits/`.

```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split

# Define splits directory
SPLITS_DIR = f"{PROJECT_ROOT}/src/data/splits"
os.makedirs(SPLITS_DIR, exist_ok=True)

train_raw_csv_path = os.path.join(RAW_DATA_DIR, "train.csv")

print("=" * 65)
print("       APTOS 2019 STRATIFIED DATASET SPLITTING (70/15/15)      ")
print("=" * 65)

if os.path.exists(train_raw_csv_path):
    df_raw = pd.read_csv(train_raw_csv_path)
    total_samples = len(df_raw)

    RANDOM_STATE = 42

    # 1. First Split: 70% Train, 30% Temp (Val + Test)
    df_train, df_temp = train_test_split(
        df_raw,
        test_size=0.30,
        random_state=RANDOM_STATE,
        stratify=df_raw["diagnosis"]
    )

    # 2. Second Split: Split Temp 50/50 into Validation (15% total) and Test (15% total)
    df_val, df_test = train_test_split(
        df_temp,
        test_size=0.50,
        random_state=RANDOM_STATE,
        stratify=df_temp["diagnosis"]
    )

    # 3. Reset dataframe indices
    df_train = df_train.reset_index(drop=True)
    df_val = df_val.reset_index(drop=True)
    df_test = df_test.reset_index(drop=True)

    # 4. Save split CSV files
    train_split_path = os.path.join(SPLITS_DIR, "train.csv")
    val_split_path = os.path.join(SPLITS_DIR, "val.csv")
    test_split_path = os.path.join(SPLITS_DIR, "test.csv")

    df_train.to_csv(train_split_path, index=False)
    df_val.to_csv(val_split_path, index=False)
    df_test.to_csv(test_split_path, index=False)

    print(f"Total Original Samples: {total_samples}")
    print("-" * 65)

    splits_info = [
        ("Train (70%)", df_train, train_split_path),
        ("Val   (15%)", df_val, val_split_path),
        ("Test  (15%)", df_test, test_split_path),
    ]

    # Print sample counts and percentages
    for name, df_split, path in splits_info:
        count = len(df_split)
        pct = (count / total_samples) * 100
        print(f"  {name:<12}: {count:>5} samples ({pct:5.2f}%) | Saved: {path}")

    print("\n" + "-" * 65)
    print("Diagnosis Class Distribution Across Splits (Counts & Percentages):")
    print("-" * 65)

    for name, df_split, _ in splits_info:
        print(f"\n  --- {name} --- ")
        counts = df_split["diagnosis"].value_counts().sort_index()
        for class_id, cnt in counts.items():
            class_pct = (cnt / len(df_split)) * 100
            print(f"    Class {class_id}: {cnt:>5} samples ({class_pct:5.2f}%)")

    # Verification Checks
    print("\n" + "=" * 65)
    print("             STRATIFIED SPLIT VERIFICATION CHECKS              ")
    print("=" * 65)

    train_ids = set(df_train["id_code"])
    val_ids = set(df_val["id_code"])
    test_ids = set(df_test["id_code"])

    # Check 1: No overlap between train, val, and test IDs
    train_val_overlap = len(train_ids.intersection(val_ids))
    train_test_overlap = len(train_ids.intersection(test_ids))
    val_test_overlap = len(val_ids.intersection(test_ids))
    no_overlap = (train_val_overlap == 0 and train_test_overlap == 0 and val_test_overlap == 0)
    print(f"  1. No ID Overlap (Train/Val/Test):          {no_overlap} (Overlaps: {train_val_overlap}/{train_test_overlap}/{val_test_overlap})")

    # Check 2: All 3662 original IDs present exactly once
    all_combined_ids = list(df_train["id_code"]) + list(df_val["id_code"]) + list(df_test["id_code"])
    unique_combined_ids = set(all_combined_ids)
    all_present_once = (len(all_combined_ids) == total_samples) and (len(unique_combined_ids) == total_samples)
    print(f"  2. All {total_samples} IDs present exactly once:     {all_present_once} (Unique IDs: {len(unique_combined_ids)})")

    # Check 3: All 5 diagnosis classes in every split
    train_classes = set(df_train["diagnosis"].unique())
    val_classes = set(df_val["diagnosis"].unique())
    test_classes = set(df_test["diagnosis"].unique())
    expected_5_classes = {0, 1, 2, 3, 4}
    all_classes_present = (train_classes == expected_5_classes and val_classes == expected_5_classes and test_classes == expected_5_classes)
    print(f"  3. All 5 classes (0-4) in every split:      {all_classes_present}")

    print("-" * 65)
    if no_overlap and all_present_once and all_classes_present:
        print("  STRATIFIED SPLIT VERIFICATION: PASS")
    else:
        print("  STRATIFIED SPLIT VERIFICATION: FAIL")
    print("=" * 65)
else:
    print(f"ERROR: 'train.csv' not found at '{train_raw_csv_path}'.")
    print("Please run previous notebook cells first.")
```

# 9. PyTorch Data Pipeline Smoke Test

Imports the data pipeline modules from `src/data/`, creates DataLoaders for train, val, and test splits, fetches a sample training batch, and verifies tensor shapes and class weights.

```python
import sys
import os
import torch
import numpy as np

# Ensure PROJECT_ROOT is set and in sys.path
if "PROJECT_ROOT" not in locals() and "PROJECT_ROOT" not in globals():
    PROJECT_ROOT = "/content/drive/MyDrive/DR_project"

# Auto-detect directory casing if needed
for c in ["/content/drive/MyDrive/DR_project", "/content/drive/MyDrive/DR_Project"]:
    if os.path.exists(c):
        PROJECT_ROOT = c
        break

if PROJECT_ROOT not in sys.path:
    sys.path.insert(0, PROJECT_ROOT)

print("=" * 65)
print("            PYTORCH DATA PIPELINE SMOKE TEST           ")
print("=" * 65)
print(f"Using PROJECT_ROOT: {PROJECT_ROOT}")

# ---- Define paths ----
SPLITS_DIR = f"{PROJECT_ROOT}/src/data/splits"
RAW_IMG_DIR = f"{PROJECT_ROOT}/src/data/raw/train_images"

# ---- Prerequisite Check: Split CSV files must exist ----
required_splits = ["train.csv", "val.csv", "test.csv"]
missing_splits = [f for f in required_splits if not os.path.isfile(os.path.join(SPLITS_DIR, f))]

if missing_splits:
    print(f"\nERROR: The following split CSV files are missing from {SPLITS_DIR}:")
    for m in missing_splits:
        print(f"  - {m}")
    print("\nPlease run Section 8 (Reproducible Stratified Dataset Splitting) first.")
    raise FileNotFoundError(f"Missing split files: {missing_splits}. Run Section 8 first.")

if not os.path.isdir(RAW_IMG_DIR):
    print(f"\nERROR: Image directory not found: {RAW_IMG_DIR}")
    print("Please run Sections 4-5 (Dataset Download & Extraction) first.")
    raise FileNotFoundError(f"Image directory not found: {RAW_IMG_DIR}")

print(f"Splits directory: {SPLITS_DIR} (all 3 CSV files found)")
print(f"Image directory:  {RAW_IMG_DIR} (exists)")

# ---- Ensure __init__.py markers exist ----
src_init = os.path.join(PROJECT_ROOT, "src", "__init__.py")
data_init = os.path.join(PROJECT_ROOT, "src", "data", "__init__.py")

if not os.path.exists(src_init):
    with open(src_init, "w") as f:
        f.write("# Package marker for src\n")

if not os.path.exists(data_init):
    with open(data_init, "w") as f:
        f.write("# Package marker for src.data\n")

# ---- Force reimport of src.data modules ----
for mod_name in list(sys.modules.keys()):
    if mod_name.startswith("src.data") or mod_name == "src":
        del sys.modules[mod_name]

from src.data.dataloader import get_dataloaders

print("\nImport: src.data.dataloader.get_dataloaders ... OK")

# ---- Determine safe num_workers setting ----
num_workers = 2 if os.name != 'nt' else 0

# ---- Create DataLoaders for train, val, and test splits ----
dataloaders = get_dataloaders(
    splits_dir=SPLITS_DIR,
    img_dir=RAW_IMG_DIR,
    batch_size=32,
    num_workers=num_workers,
    img_size=224,
    pin_memory=torch.cuda.is_available()
)

train_loader = dataloaders["train"]
val_loader = dataloaders["val"]
test_loader = dataloaders["test"]

train_size = len(train_loader.dataset)
val_size = len(val_loader.dataset)
test_size = len(test_loader.dataset)

print(f"\n1. Dataset Sizes:")
print(f"   Train set size: {train_size}")
print(f"   Val set size:   {val_size}")
print(f"   Test set size:  {test_size}")

# ---- Compute Class Weights (Inverse Class Frequency) ----
train_targets = train_loader.dataset.df["diagnosis"].values
class_counts = np.bincount(train_targets, minlength=5)
total_samples = len(train_targets)
num_classes = 5
class_weights = total_samples / (num_classes * class_counts)

print("\n2. Class Counts & Computed Class Weights:")
for class_id in range(num_classes):
    print(f"   Class {class_id}: Count = {class_counts[class_id]:>4}, Weight = {class_weights[class_id]:.4f}")

# ---- Retrieve exactly one batch from training DataLoader ----
images, labels = next(iter(train_loader))

print("\n3. Training Batch Properties:")
print(f"   Training Image Batch Shape: {list(images.shape)}")
print(f"   Training Label Batch Shape: {list(labels.shape)}")
print(f"   Training Labels:           {labels.tolist()}")

# ---- Verification Checks ----
expected_img_shape = torch.Size([32, 3, 224, 224])
expected_lbl_shape = torch.Size([32])

img_shape_valid = (images.shape == expected_img_shape)
lbl_shape_valid = (labels.shape == expected_lbl_shape)

print("\n" + "=" * 65)
print("                   SMOKE TEST VERIFICATION             ")
print("=" * 65)
print(f"  Image batch shape is [32, 3, 224, 224]:  {img_shape_valid} (Actual: {list(images.shape)})")
print(f"  Label batch shape is [32]:               {lbl_shape_valid} (Actual: {list(labels.shape)})")
print("-" * 65)

if img_shape_valid and lbl_shape_valid:
    print("  DATA PIPELINE SMOKE TEST: PASS")
else:
    print("  DATA PIPELINE SMOKE TEST: FAIL")
print("=" * 65)
```
