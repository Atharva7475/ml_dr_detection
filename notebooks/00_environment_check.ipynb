# 4. PyTorch Verification

# 5. Core Package Verification

```python
packages_to_check = [
    ("numpy", "numpy"),
    ("pandas", "pandas"),
    ("sklearn", "scikit-learn"),
    ("matplotlib", "matplotlib"),
    ("seaborn", "seaborn"),
    ("PIL", "Pillow"),
    ("cv2", "opencv-python"),
    ("tqdm", "tqdm"),
    ("yaml", "PyYAML"),
    ("scipy", "scipy"),
    ("albumentations", "albumentations")
]

package_status = {}

print("Checking required package imports:\n" + "-"*40)
for module_name, display_name in packages_to_check:
    try:
        mod = __import__(module_name)
        ver = getattr(mod, "__version__", "Installed")
        print(f"  [PASS] {display_name} ({module_name}): v{ver}")
        package_status[display_name] = True
    except ImportError as e:
        print(f"  [FAIL] {display_name} ({module_name}): NOT INSTALLED ({e})")
        package_status[display_name] = False
```

# 6. Final Environment Status

```python
import sys
import os

print("=" * 45)
print("       FINAL ENVIRONMENT STATUS SUMMARY      ")
print("=" * 45)

python_status = "PASS" if sys.version_info >= (3, 7) else "FAIL"
drive_status = "PASS" if os.path.exists("/content/drive") else "FAIL"
dir_status = "PASS" if os.path.exists(PROJECT_ROOT) else "FAIL"
pytorch_status = "PASS" if 'torch' in sys.modules else "FAIL"
cuda_status = "AVAILABLE" if torch.cuda.is_available() else "NOT AVAILABLE"
gpu_status = f"detected ({torch.cuda.get_device_name(0)})" if torch.cuda.is_available() else "not detected"
all_pkgs_pass = all(package_status.values()) if 'package_status' in locals() else False
pkgs_status_str = "PASS" if all_pkgs_pass else "FAIL"

print(f"Python:             {python_status}")
print(f"Google Drive:       {drive_status}")
print(f"Project directory:  {dir_status}")
print(f"PyTorch:            {pytorch_status}")
print(f"CUDA:               {cuda_status}")
print(f"GPU:                {gpu_status}")
print(f"Required packages:  {pkgs_status_str}")
print("=" * 45)
```
